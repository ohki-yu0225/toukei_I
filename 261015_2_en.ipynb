### Loading the data

Source: Japan Statistical Society (ed.) (2015). *Revised Edition: Japan Statistical Society Official Certification, Statistics Test Grade 2 — Basics of Statistics*. Tokyo Tosho. (in Japanese)

The columns are: `Proximity` (closeness to the station, A or B), `Rent` (monthly rent in yen), `Layout` (floor plan), `Size` (floor area in m²), `Direction` (the direction the room faces), and `Age` (building age in years).

```python
import gdown
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (4, 3)

file_id = "17rKrgPe8D2LadSw1YQkOVp3YvTeiT0ak"

gdown.download(
    id=file_id,
    output="room.csv",
    quiet=False
)

df = pd.read_csv("room.csv", encoding="shift-jis")
df.columns = ["Proximity", "Rent", "Layout", "Size", "Direction", "Age"]
df["Direction"] = df["Direction"].replace({
    "北": "N", "北東": "NE", "東": "E", "南東": "SE",
    "南": "S", "南西": "SW", "西": "W", "北西": "NW"
})
df
```

### Summarizing a qualitative variable

To summarize a qualitative variable, we count the number of observations that fall into each category (called the frequency).

```python
freq = df["Direction"].value_counts()
rel_freq = df["Direction"].value_counts(normalize=True).round(3)
freq_table = pd.concat([freq, rel_freq], axis=1)
freq_table.columns = ["Frequency", "Relative frequency"]
freq_table
```

The result of summarizing a qualitative variable can be visualized with a bar chart.

```python
plt.bar(freq.index, freq, edgecolor="black")
plt.xlabel("Direction")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.bar(rel_freq.index, rel_freq, edgecolor="black")
plt.xlabel("Direction")
plt.ylabel("Relative frequency")
plt.show()
```

### Frequency distribution table

Dividing the values of a quantitative variable into several groups (classes) and counting them is called a frequency distribution.

```python
bins = range(60000, 150001, 10000)
rent_freq, _ = np.histogram(df["Rent"], bins=bins)

labels = [f"{b}-{b + 10000}" for b in bins[:-1]]

rent_table = pd.DataFrame({"Frequency": rent_freq, "Relative frequency": (rent_freq / rent_freq.sum()).round(3)}, index=labels)
rent_table
```

### Histogram

A histogram is one way to graph a frequency distribution. The horizontal axis shows the values of the variable, and over each class interval a rectangle is drawn so that its area is proportional to the frequency.

```python
plt.hist(df["Rent"], bins=bins, edgecolor="black")
plt.xlabel("Rent (yen)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(df["Rent"], bins=bins, weights=np.ones(len(df)) / len(df), edgecolor="black")
plt.xlabel("Rent (yen)")
plt.ylabel("Relative frequency")
plt.show()
```

### Understanding the shape of a distribution

The purpose of a histogram is to understand the features of the distribution of a quantitative variable. Specifically, it shows where the center of the distribution is, how widely the values are spread, and whether the distribution is symmetric or has a long tail on one side.

```python
np.random.seed(0)
left = np.random.beta(5, 2, 1000) * 100
sym = np.random.beta(5, 5, 1000) * 100
right = np.random.beta(2, 5, 1000) * 100

fig, axes = plt.subplots(1, 3, figsize=(10, 3), sharex=True, sharey=True)
axes[0].hist(left, bins=10, edgecolor="black")
axes[0].set_title("Long left tail")
axes[1].hist(sym, bins=10, edgecolor="black")
axes[1].set_title("Symmetric")
axes[2].hist(right, bins=10, edgecolor="black")
axes[2].set_title("Long right tail")
plt.show()
```
