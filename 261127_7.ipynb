```python
!pip install japanize-matplotlib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import japanize_matplotlib
from scipy import stats as st

plt.rcParams["figure.figsize"] = (4, 3)
%config InlineBackend.figure_format = "retina"
```

```python
# サイコロの出る目の確率分布を定義
x = np.arange(1, 7)          # 出る目 1〜6
p = np.full(6, 1/6)          # 各目の確率（一様）
df = pd.DataFrame({"x": x, "p": p})
print("確率の合計:", p.sum())
df
```

```python
# 確率分布を棒グラフで可視化
fig, ax = plt.subplots()
ax.bar(x, p)
ax.set_xlabel("出る目 x")
ax.set_ylabel("確率 p(x)")
ax.set_title("歪みのないサイコロの確率分布")
ax.set_xticks(x)
ax.set_ylim(0, 0.5)
plt.show()
```

```python
# 平均・分散・標準偏差を計算
mean = np.sum(x * p)                 # E[X] = Σ x p(x)
var = np.sum((x - mean) ** 2 * p)    # V[X] = Σ (x-μ)^2 p(x)
std = np.sqrt(var)
print(f"平均: {mean:.4f}")
print(f"分散: {var:.4f}")
print(f"標準偏差: {std:.4f}")
```

```python
# 歪みのあるサイコロの出る目の確率分布を定義
p_center = np.array([1/12, 1/12, 1/3, 1/3, 1/12, 1/12])
p_edge   = np.array([1/3, 1/12, 1/12, 1/12, 1/12, 1/3])
df_skew = pd.DataFrame({"x": x, "p_center": p_center, "p_edge": p_edge})
print("確率の合計:", p_center.sum(), p_edge.sum())
df_skew
```

```python
# 確率分布を棒グラフで可視化
fig, axes = plt.subplots(1, 2, figsize=(8, 3), sharey=True)
for ax, prob, title in zip(axes, [p_center, p_edge], ["3,4が出やすいサイコロ", "1,6が出やすいサイコロ"]):
    ax.bar(x, prob)
    ax.set_xlabel("出る目 x")
    ax.set_title(title)
    ax.set_xticks(x)
    ax.set_ylim(0, 0.5)
axes[0].set_ylabel("確率 p(x)")
plt.tight_layout()
plt.show()
```

```python
# 平均・分散・標準偏差を計算
def calc_stats(x, p):
    mean = np.sum(x * p)
    var = np.sum((x - mean) ** 2 * p)
    return mean, var, np.sqrt(var)

stats = pd.DataFrame(
    [calc_stats(x, p), calc_stats(x, p_center), calc_stats(x, p_edge)],
    index=["歪みのない", "3,4が出やすい", "1,6が出やすい"],
    columns=["平均", "分散", "標準偏差"],
)
stats.round(4)
```

```python
# N=10, p=0.3の二項分布の確率分布を定義
N, p = 10, 0.3
x = np.arange(0, N + 1)                   # 取りうる値 0〜N
f = st.binom.pmf(x, N, p)                 # 確率関数 p(x)
df = pd.DataFrame({"x": x, "p(x)": f})
df.round(4)
```

```python
# 確率関数を図示
fig, ax = plt.subplots()
ax.bar(x, f)
ax.set_xlabel("x")
ax.set_ylabel("確率 p(x)")
ax.set_title(f"二項分布 B({N}, {p})")
ax.set_xticks(x)
plt.show()
```

```python
# p=0.1,0.3,0.5の二項分布の確率関数を図示（マーカー付き折線にする）
fig, ax = plt.subplots()
for prob in [0.1, 0.3, 0.5]:              # 次のセルでN, pを使うので，pは上書きしない
    ax.plot(x, st.binom.pmf(x, N, prob), marker="o", label=f"p={prob}")
ax.set_xlabel("x")
ax.set_ylabel("確率 p(x)")
ax.set_title(f"二項分布 B({N}, p)")
ax.set_xticks(x)
ax.legend()
plt.show()
```

```python
# 平均・分散・標準偏差を計算
mean = st.binom.mean(N, p)
var  = st.binom.var(N, p)
std  = st.binom.std(N, p)
print(f"平均: {mean:.4f}")
print(f"分散: {var:.4f}")
print(f"標準偏差: {std:.4f}")
```

```python
# lambda=2のポアソン分布の確率分布を定義
lam = 2
x_max = 10                                # 本来は0〜∞だが，10で打ち切る（表・図の上限）

x = np.arange(0, x_max + 1)
f = st.poisson.pmf(x, lam)                # 確率関数 p(x)
df = pd.DataFrame({"x": x, "p(x)": f})
df.round(4)
```

```python
# 確率関数を図示
fig, ax = plt.subplots()
ax.bar(x, f)
ax.set_xlabel("x")
ax.set_ylabel("確率 p(x)")
ax.set_title(f"ポアソン分布 Po({lam})")
ax.set_xticks(x)
ax.set_xlim(-0.5, x_max + 0.5)
ax.set_ylim(0, 0.4)
plt.show()
```

```python
# 平均・分散・標準偏差を計算
mean = st.poisson.mean(lam)
var  = st.poisson.var(lam)
std  = st.poisson.std(lam)
print(f"平均: {mean:.4f}")
print(f"分散: {var:.4f}")
print(f"標準偏差: {std:.4f}")
```

```python
# lambda=1,2,3のポアソン分布の確率関数を図示（マーカー付き折線にする）
fig, ax = plt.subplots()
for l in [1, 2, 3]:                       # 上の図と同じ範囲 0〜x_max で描画
    ax.plot(x, st.poisson.pmf(x, l), marker="o", label=f"λ={l}")
ax.set_xlabel("x")
ax.set_ylabel("確率 p(x)")
ax.set_title("ポアソン分布 Po(λ)")
ax.set_xticks(x)
ax.set_xlim(-0.5, x_max + 0.5)
ax.set_ylim(0, 0.4)
ax.legend()
plt.show()
```

```python
# 演習：二項分布のパラメータを変えて確率関数を図示する
N, p = 10, 0.3                            # 試行回数・成功確率（ここを変える）

# 確率分布の表
x = np.arange(0, N + 1)
f = st.binom.pmf(x, N, p)
df = pd.DataFrame({"x": x, "p(x)": f})

fig, ax = plt.subplots()
ax.bar(x, f)
ax.set_xlabel("x")
ax.set_ylabel("確率 p(x)")
ax.set_title(f"二項分布 B({N}, {p})")
plt.show()

# 平均・分散・標準偏差を計算
mean = st.binom.mean(N, p)
var  = st.binom.var(N, p)
std  = st.binom.std(N, p)
print(f"平均: {mean:.4f}")
print(f"分散: {var:.4f}")
print(f"標準偏差: {std:.4f}")

df.round(4)
```

```python
# 演習：ポアソン分布のパラメータを変えて確率関数を図示する
lam = 1                                   # λ（ここを変える）

# 上側確率 P(X > x) が alpha 以下になる最小の x を描画・表の上限とする
alpha = 0.001                             # 上側確率 P(X > x_max) の基準
x_max = int(st.poisson.isf(alpha, lam))

# 確率分布の表
x = np.arange(0, x_max + 1)
f = st.poisson.pmf(x, lam)
df = pd.DataFrame({"x": x, "p(x)": f})

fig, ax = plt.subplots()
ax.bar(x, f)
ax.set_xlabel("x")
ax.set_ylabel("確率 p(x)")
ax.set_title(f"ポアソン分布 Po({lam})")
plt.show()

# 平均・分散・標準偏差を計算
mean = st.poisson.mean(lam)
var  = st.poisson.var(lam)
std  = st.poisson.std(lam)
print(f"平均: {mean:.4f}")
print(f"分散: {var:.4f}")
print(f"標準偏差: {std:.4f}")

df.round(4)
```
