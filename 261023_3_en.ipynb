```python
import gdown
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (4, 3)
%config InlineBackend.figure_format = "retina"

file_id = "17rKrgPe8D2LadSw1YQkOVp3YvTeiT0ak"

gdown.download(
    id=file_id,
    output="room.csv",
    quiet=False
)

df = pd.read_csv("room.csv", encoding="shift-jis")
df.columns = ["Proximity", "Rent", "Layout", "Size", "Direction", "Age"]
df["Direction"] = df["Direction"].replace({
    "北": "N", "北東": "NE", "東": "E", "南東": "SE",
    "南": "S", "南西": "SW", "西": "W", "北西": "NW"
})
df
```

### Mean, Standard Deviation, and Variance

The mean is a widely used measure of central tendency. It is defined as the sum of the observations divided by the number of observations. The standard deviation and variance are measures of how widely the data are spread around the mean.


```python
mean = df["Rent"].mean()
std = df["Rent"].std()
var = df["Rent"].var()

print("Mean:", round(mean, 1), "yen")
print("Standard deviation:", round(std, 1), "yen")
print("Variance:", round(var, 1), "yen^2")
```

```python
bins = range(60000, 150001, 10000)

plt.figure(figsize=(5, 3))
plt.hist(df["Rent"], bins=bins, color="#c8c8c8", edgecolor="black")
plt.axvline(mean, color="#c81e1e", label="Mean", linewidth=2)
plt.axvline(mean - std, color="#0050b3", linestyle="--", label="Mean ± SD", linewidth=2)
plt.axvline(mean + std, color="#0050b3", linestyle="--", linewidth=2)
plt.xlabel("Rent (yen)")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

### Median and Mode

The mean is easily affected by outliers. For this reason, other measures of central tendency, such as the median and the mode, are sometimes used instead.

```python
median = df["Rent"].median()

counts, edges = np.histogram(df["Rent"], bins=bins)
i = counts.argmax()
mode = (edges[i] + edges[i + 1]) / 2

print("Median:", round(median, 1), "yen")
print("Mode:", round(mode, 1), "yen")
```

### Range and Interquartile Range

When the observations are sorted in ascending order, the values located at each quarter are called the first quartile, the second quartile (equal to the median), and the third quartile. The interquartile range is the difference between the third and first quartiles, and is one measure of spread.

```python
q1 = df["Rent"].quantile(0.25)
q2 = df["Rent"].quantile(0.50)
q3 = df["Rent"].quantile(0.75)
iqr = q3 - q1

print("First quartile:", round(q1, 1), "yen")
print("Median (second quartile):", round(q2, 1), "yen")
print("Third quartile:", round(q3, 1), "yen")
print("Interquartile range:", round(iqr, 1), "yen")
```

```python
plt.figure(figsize=(5, 3))
plt.hist(df["Rent"], bins=bins, color="#c8c8c8", edgecolor="black")
plt.axvline(q1, color="#0050b3", linestyle="--", label="First / third quartiles", linewidth=2)
plt.axvline(q2, color="#c81e1e", label="Median (second quartile)", linewidth=2)
plt.axvline(q3, color="#0050b3", linestyle="--", linewidth=2)
plt.xlabel("Rent (yen)")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

### Exercise: Measures of Central Tendency in a Right-Skewed Distribution

In a right-skewed distribution, the mean, median, and mode diverge, and they usually follow the order mode < median < mean. Change the value of `skewness` (the parameter that controls the length of the right tail) in the code below, run it, and see how the differences among the three measures change as the right tail gets longer.

```python
# Try changing this: length of the right tail (skewness). 0 = symmetric; larger = longer right tail (0 to 2)
skewness = 0

n = 10000

# Symmetric distribution: normal distribution with mean 50 and SD 15
np.random.seed(0)
sym = np.random.normal(50, 15, n)

# Right-skewed distribution: normal plus exponential (mean 50, SD 15)
tau = 15 * (skewness / 2) ** (1 / 3)
sigma = np.sqrt(15**2 - tau**2)
skew = np.random.normal(50 - tau, sigma, n) + np.random.exponential(tau, n)

bins2 = np.arange(-25, 206, 10)

# Mode (midpoint of the class with the highest frequency)
counts, edges = np.histogram(sym, bins=bins2)
i = counts.argmax()
sym_mode = (edges[i] + edges[i + 1]) / 2

counts, edges = np.histogram(skew, bins=bins2)
i = counts.argmax()
skew_mode = (edges[i] + edges[i + 1]) / 2

print("skewness =", skewness)
print("Symmetric     mean:", round(np.mean(sym), 1), " median:", round(np.median(sym), 1), " mode:", round(sym_mode, 1))
print("Right-skewed  mean:", round(np.mean(skew), 1), " median:", round(np.median(skew), 1), " mode:", round(skew_mode, 1))

plt.figure(figsize=(8, 3))

plt.subplot(1, 2, 1)
plt.hist(sym, bins=bins2, color="#c8c8c8", edgecolor="black")
plt.axvline(np.mean(sym), color="#c81e1e", label="Mean", linewidth=1.5)
plt.axvline(np.median(sym), color="#0050b3", linestyle="--", label="Median", linewidth=1.5)
plt.axvline(sym_mode, color="#00754e", linestyle=":", label="Mode", linewidth=1.5)
plt.title("Symmetric distribution")
plt.legend()

plt.subplot(1, 2, 2)
plt.hist(skew, bins=bins2, color="#c8c8c8", edgecolor="black")
plt.axvline(np.mean(skew), color="#c81e1e", label="Mean", linewidth=1.5)
plt.axvline(np.median(skew), color="#0050b3", linestyle="--", label="Median", linewidth=1.5)
plt.axvline(skew_mode, color="#00754e", linestyle=":", label="Mode", linewidth=1.5)
plt.title("Right-skewed distribution (skewness=" + str(skewness) + ")")
plt.legend()

plt.tight_layout()
plt.show()
```
