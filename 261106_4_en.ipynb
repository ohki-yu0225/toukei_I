```python
import gdown
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (4, 3)
%config InlineBackend.figure_format = "retina"

file_id = "17rKrgPe8D2LadSw1YQkOVp3YvTeiT0ak"

gdown.download(
    id=file_id,
    output="room.csv",
    quiet=False
)

df = pd.read_csv("room.csv", encoding="shift-jis")
df.columns = ["Proximity", "Rent", "Layout", "Size", "Direction", "Age"]
df["Direction"] = df["Direction"].replace({
    "北": "N", "北東": "NE", "東": "E", "南東": "SE",
    "南": "S", "南西": "SW", "西": "W", "北西": "NW"
})
df
```

### Scatter Plot

The relationship between two quantitative variables can be visualized with a scatter plot, in which the values of the two variables are mapped to the 𝑥-axis and the 𝑦-axis.

```python
# Draw a scatter plot of room size and rent
plt.scatter(df["Size"], df["Rent"], color="#0050b3", alpha=1, s=10)
plt.xlabel("Size (m²)")
plt.ylabel("Rent (yen)")
plt.show()
```

### Correlation Coefficient

The correlation coefficient is a measure that expresses the strength of the correlation between two variables on a scale from -1 to 1.

```python
# Calculate the correlation coefficient between room size and rent
correlation = np.corrcoef(df["Size"], df["Rent"])[0, 1]
print(f"Correlation coefficient: {correlation:.2f}")
```

### Stratified Scatter Plot

When group information is available, scatter plots are sometimes drawn separately for each group.


```python
# Stratify the scatter plot of room size and rent by proximity
room_A = df[df["Proximity"] == "A"]
room_B = df[df["Proximity"] == "B"]

plt.scatter(room_A["Size"], room_A["Rent"], color="#0050b3", marker="o", s=10, label="Near")
plt.scatter(room_B["Size"], room_B["Rent"], color="#c81e1e", marker="^", s=12, label="Far")
plt.xlabel("Size (m²)")
plt.ylabel("Rent (yen)")
plt.legend(loc="lower right", frameon=False)
plt.show()
```

### Regression Line

When a straight line is used to predict the relationship between variables, this line is called a regression line.

```python
# Draw a regression line on the scatter plot of room size and rent
a, b = np.polyfit(df["Size"], df["Rent"], 1)
print("Slope:", round(a, 1), "yen/m²")
print("Intercept:", round(b, 1), "yen")

xs = np.linspace(df["Size"].min(), df["Size"].max(), 100)

plt.scatter(df["Size"], df["Rent"], color="#0050b3", alpha=1, s=10)
plt.plot(xs, a * xs + b, color="#c81e1e", linewidth=2)
plt.xlabel("Size (m²)")
plt.ylabel("Rent (yen)")
plt.show()
```

```python
# Caution: The correlation coefficient only captures linear relationships
n = 100

np.random.seed(103)
x = np.random.normal(0, 1, n)
e = np.random.normal(0, 1, n)

# Standardization: subtract the mean and divide by the standard deviation (mean 0, SD 1)
x = (x - x.mean()) / x.std()

# Left: data from y = r * x + sqrt(1 - r^2) * e with r = 0 (no relationship)
y1 = 0 * x + np.sqrt(1 - 0**2) * e
y1 = (y1 - y1.mean()) / y1.std()

# Right: data on the quadratic function y = x^2 (a clear relationship)
y2 = x**2 + 0.2 * e
y2 = (y2 - y2.mean()) / y2.std()

r1 = np.corrcoef(x, y1)[0, 1]
r2 = np.corrcoef(x, y2)[0, 1]

plt.figure(figsize=(6, 3))

plt.subplot(1, 2, 1)
plt.scatter(x, y1, color="#0050b3", s=10)
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.xlabel("x")
plt.ylabel("y")
plt.title(f"r = {r1:.2f}")

plt.subplot(1, 2, 2)
plt.scatter(x, y2, color="#0050b3", s=10)
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.xlabel("x")
plt.ylabel("y")
plt.title(f"r = {r2:.2f}")

plt.tight_layout()
plt.show()
```

```python
# Caution: The correlation coefficient is easily affected by outliers
n = 50

np.random.seed(0)
x = np.random.normal(0, 1, n)
e = np.random.normal(0, 1, n)

# Standardization: subtract the mean and divide by the standard deviation (mean 0, SD 1)
x = (x - x.mean()) / x.std()

# Left: data from y = r * x + sqrt(1 - r^2) * e with r = -0.4 (negative correlation)
y = -0.4 * x + np.sqrt(1 - 0.4**2) * e
y = (y - y.mean()) / y.std()

# Right: add 5 outliers in the upper right
x_out = np.array([3.0, 3.2, 3.4, 3.6, 3.8])
y_out = np.array([3.5, 3.1, 3.7, 3.3, 3.6])
x_all = np.concatenate([x, x_out])
y_all = np.concatenate([y, y_out])

r1 = np.corrcoef(x, y)[0, 1]
r2 = np.corrcoef(x_all, y_all)[0, 1]

plt.figure(figsize=(6, 3))

plt.subplot(1, 2, 1)
plt.scatter(x, y, color="#0050b3", s=10)
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.xlabel("x")
plt.ylabel("y")
plt.title(f"r = {r1:.2f}")

plt.subplot(1, 2, 2)
plt.scatter(x, y, color="#0050b3", s=10)
plt.scatter(x_out, y_out, color="#c81e1e", s=10)
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.xlabel("x")
plt.ylabel("y")
plt.title(f"r = {r2:.2f}")

plt.tight_layout()
plt.show()
```
