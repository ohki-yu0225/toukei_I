```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats as st

plt.rcParams["figure.figsize"] = (4, 3)
%config InlineBackend.figure_format = "retina"
```

```python
# Define the probability distribution of a die roll
x = np.arange(1, 7)          # faces 1-6
p = np.full(6, 1/6)          # probability of each face (uniform)
df = pd.DataFrame({"x": x, "p": p})
print("Sum of probabilities:", p.sum())
df
```

```python
# Visualize the probability distribution as a bar chart
fig, ax = plt.subplots()
ax.bar(x, p)
ax.set_xlabel("Face x")
ax.set_ylabel("Probability p(x)")
ax.set_title("Probability distribution of a fair die")
ax.set_xticks(x)
ax.set_ylim(0, 0.5)
plt.show()
```

```python
# Compute the mean, variance and standard deviation
mean = np.sum(x * p)                 # E[X] = Σ x p(x)
var = np.sum((x - mean) ** 2 * p)    # V[X] = Σ (x-μ)^2 p(x)
std = np.sqrt(var)
print(f"Mean: {mean:.4f}")
print(f"Variance: {var:.4f}")
print(f"Standard deviation: {std:.4f}")
```

```python
# Define the probability distributions of biased dice
p_center = np.array([1/12, 1/12, 1/3, 1/3, 1/12, 1/12])
p_edge   = np.array([1/3, 1/12, 1/12, 1/12, 1/12, 1/3])
df_skew = pd.DataFrame({"x": x, "p_center": p_center, "p_edge": p_edge})
print("Sum of probabilities:", p_center.sum(), p_edge.sum())
df_skew
```

```python
# Visualize the probability distributions as bar charts
fig, axes = plt.subplots(1, 2, figsize=(8, 3), sharey=True)
for ax, prob, title in zip(axes, [p_center, p_edge], ["Die favoring 3 and 4", "Die favoring 1 and 6"]):
    ax.bar(x, prob)
    ax.set_xlabel("Face x")
    ax.set_title(title)
    ax.set_xticks(x)
    ax.set_ylim(0, 0.5)
axes[0].set_ylabel("Probability p(x)")
plt.tight_layout()
plt.show()
```

```python
# Compute the mean, variance and standard deviation
def calc_stats(x, p):
    mean = np.sum(x * p)
    var = np.sum((x - mean) ** 2 * p)
    return mean, var, np.sqrt(var)

stats = pd.DataFrame(
    [calc_stats(x, p), calc_stats(x, p_center), calc_stats(x, p_edge)],
    index=["Fair", "Favoring 3 and 4", "Favoring 1 and 6"],
    columns=["Mean", "Variance", "Std. dev."],
)
stats.round(4)
```

```python
# Define the binomial distribution with N=10, p=0.3
N, p = 10, 0.3
x = np.arange(0, N + 1)                   # possible values 0-N
f = st.binom.pmf(x, N, p)                 # probability mass function p(x)
df = pd.DataFrame({"x": x, "p(x)": f})
df.round(4)
```

```python
# Plot the probability mass function
fig, ax = plt.subplots()
ax.bar(x, f)
ax.set_xlabel("x")
ax.set_ylabel("Probability p(x)")
ax.set_title(f"Binomial distribution B({N}, {p})")
ax.set_xticks(x)
plt.show()
```

```python
# Plot the PMFs of binomial distributions with p=0.1, 0.3, 0.5 (lines with markers)
fig, ax = plt.subplots()
for prob in [0.1, 0.3, 0.5]:              # N and p are used in the next cell, so do not overwrite p
    ax.plot(x, st.binom.pmf(x, N, prob), marker="o", label=f"p={prob}")
ax.set_xlabel("x")
ax.set_ylabel("Probability p(x)")
ax.set_title(f"Binomial distribution B({N}, p)")
ax.set_xticks(x)
ax.legend()
plt.show()
```

```python
# Compute the mean, variance and standard deviation
mean = st.binom.mean(N, p)
var  = st.binom.var(N, p)
std  = st.binom.std(N, p)
print(f"Mean: {mean:.4f}")
print(f"Variance: {var:.4f}")
print(f"Standard deviation: {std:.4f}")
```

```python
# Define the Poisson distribution with lambda=2
lam = 2
x_max = 10                                # the support is 0 to ∞, but truncate at 10 (upper limit for table and plots)

x = np.arange(0, x_max + 1)
f = st.poisson.pmf(x, lam)                # probability mass function p(x)
df = pd.DataFrame({"x": x, "p(x)": f})
df.round(4)
```

```python
# Plot the probability mass function
fig, ax = plt.subplots()
ax.bar(x, f)
ax.set_xlabel("x")
ax.set_ylabel("Probability p(x)")
ax.set_title(f"Poisson distribution Po({lam})")
ax.set_xticks(x)
ax.set_xlim(-0.5, x_max + 0.5)
ax.set_ylim(0, 0.4)
plt.show()
```

```python
# Compute the mean, variance and standard deviation
mean = st.poisson.mean(lam)
var  = st.poisson.var(lam)
std  = st.poisson.std(lam)
print(f"Mean: {mean:.4f}")
print(f"Variance: {var:.4f}")
print(f"Standard deviation: {std:.4f}")
```

```python
# Plot the PMFs of Poisson distributions with lambda=1, 2, 3 (lines with markers)
fig, ax = plt.subplots()
for l in [1, 2, 3]:                       # plot over the same range 0-x_max as the figure above
    ax.plot(x, st.poisson.pmf(x, l), marker="o", label=f"λ={l}")
ax.set_xlabel("x")
ax.set_ylabel("Probability p(x)")
ax.set_title("Poisson distribution Po(λ)")
ax.set_xticks(x)
ax.set_xlim(-0.5, x_max + 0.5)
ax.set_ylim(0, 0.4)
ax.legend()
plt.show()
```
