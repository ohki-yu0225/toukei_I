## Simpson's Paradox: Correlation and Causation
Simulated data on exercise time and disease incidence (reference: https://www.saishunkansys.com/blog/marketing/simpson_paradox/ ).

```python
import gdown
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["figure.figsize"] = (4, 3)
%config InlineBackend.figure_format = "retina"
```

```python
rng = np.random.default_rng(0)
n = 200

age = rng.integers(20, 81, n)  # integer ages
group = np.where(age >= 50, "50 and over", "Under 50")
exercise = np.clip(2 + 3 * (age >= 50) + rng.normal(0, 1, n), 0, None)
disease = 10 + 0.15 * age - 1.0 * exercise + rng.normal(0, 2, n)

df = pd.DataFrame({"Age": age, "Age group": group,
                   "Exercise time": exercise, "Disease rate": disease})
df.head().round(1)
```

### Scatter plot and correlation for all data

```python
r_all = df["Exercise time"].corr(df["Disease rate"])
print(f"Overall correlation: r = {r_all:.3f}")

fig, ax = plt.subplots(figsize=(6, 5))
sns.regplot(data=df, x="Exercise time", y="Disease rate", ax=ax, ci=None,
            scatter_kws={"color": "gray", "alpha": 0.6}, line_kws={"color": "red"})
ax.set_title(f"Exercise time vs. disease rate (r = {r_all:.2f})")
ax.set_xlabel("Exercise time (hours/week)")
ax.set_ylabel("Disease rate (%)")
plt.tight_layout()
plt.show()
```

### Scatter plot colored by age group

```python
for g, d in df.groupby("Age group"):
    print(f"Correlation for {g}: r = {d['Exercise time'].corr(d['Disease rate']):.3f}")

palette = {"Under 50": "tab:blue", "50 and over": "orange"}
fig, ax = plt.subplots(figsize=(6, 5))
for g, d in df.groupby("Age group"):
    r = d["Exercise time"].corr(d["Disease rate"])
    sns.regplot(data=d, x="Exercise time", y="Disease rate", ax=ax, ci=None,
                color=palette[g], label=f"{g} (r = {r:.2f})",
                scatter_kws={"alpha": 0.7})
# Regression line for all data (for comparison)
sns.regplot(data=df, x="Exercise time", y="Disease rate", ax=ax, ci=None, scatter=False,
            line_kws={"color": "red", "linestyle": "--"}, label=f"All (r = {r_all:.2f})")
ax.set_title("Colored by age group")
ax.set_xlabel("Exercise time (hours/week)")
ax.set_ylabel("Disease rate (%)")
ax.legend()
plt.tight_layout()
plt.show()
```
