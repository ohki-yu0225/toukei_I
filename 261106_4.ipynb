```python
!pip install japanize-matplotlib

import gdown
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

plt.rcParams["figure.figsize"] = (4, 3)
%config InlineBackend.figure_format = "retina"

file_id = "17rKrgPe8D2LadSw1YQkOVp3YvTeiT0ak"

gdown.download(
    id=file_id,
    output="room.csv",
    quiet=False
)

df = pd.read_csv("room.csv", encoding="shift-jis")
df
```

### 散布図

2つの量的変数の関係は𝑥軸と𝑦軸に2つの変数の数値を対応させた散布図で可視化できる

```python
# 部屋の大きさと家賃の散布図を作る
plt.scatter(df["大きさ"], df["家賃"], color="#0050b3", alpha=1, s=10)
plt.xlabel("大きさ（m²）")
plt.ylabel("家賃（円）")
plt.show()
```

### 相関係数

相関係数は2変数の相関関係の強さを-1から1で表す指標である

```python
# 部屋の大きさと家賃の相関係数を計算する
correlation = np.corrcoef(df["大きさ"], df["家賃"])[0, 1]
print(f"相関係数: {correlation:.2f}")
```

### 層別散布図

グループの情報が得られるときには，グループごとに散布図を描くことがある


```python
# 部屋の大きさと家賃の散布図を近さで層別化する
room_A = df[df["近さ"] == "A"]
room_B = df[df["近さ"] == "B"]

plt.scatter(room_A["大きさ"], room_A["家賃"], color="#0050b3", marker="o", s=10, label="近い")
plt.scatter(room_B["大きさ"], room_B["家賃"], color="#c81e1e", marker="^", s=12, label="遠い")
plt.xlabel("大きさ（m²）")
plt.ylabel("家賃（円）")
plt.legend(loc="lower right", frameon=False)
plt.show()
```

### 回帰直線

直線の式を用いて変数間の関係性を予測するときに，この直線を回帰直線という

```python
# 部屋の大きさと家賃の散布図に回帰直線を引く
a, b = np.polyfit(df["大きさ"], df["家賃"], 1)
print("傾き:", round(a, 1), "円/m²")
print("切片:", round(b, 1), "円")

xs = np.linspace(df["大きさ"].min(), df["大きさ"].max(), 100)

plt.scatter(df["大きさ"], df["家賃"], color="#0050b3", alpha=1, s=10)
plt.plot(xs, a * xs + b, color="#c81e1e", linewidth=2)
plt.xlabel("大きさ（m²）")
plt.ylabel("家賃（円）")
plt.show()
```

```python
# 演習：相関係数を変えて，散布図を描く
# 実行すると入力欄が出るので，相関係数（-1以上1以下）を入力してみよう
rho = float(input("相関係数（-1以上1以下）を入力してください: "))

n = 100

np.random.seed(0)
x = np.random.normal(0, 1, n)
e = np.random.normal(0, 1, n)

# 標準化：平均を引いて標準偏差で割る（平均0，標準偏差1になる）
x = (x - x.mean()) / x.std()

# y = rho * x + sqrt(1 - rho^2) * e とすると，x と y の相関係数はおよそ rho になる
y = rho * x + np.sqrt(1 - rho**2) * e
y = (y - y.mean()) / y.std()

plt.figure()
plt.scatter(x, y, color="#0050b3", s=10)
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.xlabel("x")
plt.ylabel("y")
plt.title(f"r = {rho}")
plt.show()
```

```python
# 注意点: 相関係数は線形の関係しか示さない
n = 100

np.random.seed(103)
x = np.random.normal(0, 1, n)
e = np.random.normal(0, 1, n)

# 標準化：平均を引いて標準偏差で割る（平均0，標準偏差1になる）
x = (x - x.mean()) / x.std()

# 左：y = r * x + sqrt(1 - r^2) * e で r = 0 としたデータ（関係がない）
y1 = 0 * x + np.sqrt(1 - 0**2) * e
y1 = (y1 - y1.mean()) / y1.std()

# 右：2次関数 y = x^2 の上にあるデータ（はっきりした関係がある）
y2 = x**2 + 0.2 * e
y2 = (y2 - y2.mean()) / y2.std()

r1 = np.corrcoef(x, y1)[0, 1]
r2 = np.corrcoef(x, y2)[0, 1]

plt.figure(figsize=(6, 3))

plt.subplot(1, 2, 1)
plt.scatter(x, y1, color="#0050b3", s=10)
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.xlabel("x")
plt.ylabel("y")
plt.title(f"r = {r1:.2f}")

plt.subplot(1, 2, 2)
plt.scatter(x, y2, color="#0050b3", s=10)
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.xlabel("x")
plt.ylabel("y")
plt.title(f"r = {r2:.2f}")

plt.tight_layout()
plt.show()
```

```python
# 注意点: 相関係数は外れ値の影響を受けやすい
n = 50

np.random.seed(0)
x = np.random.normal(0, 1, n)
e = np.random.normal(0, 1, n)

# 標準化：平均を引いて標準偏差で割る（平均0，標準偏差1になる）
x = (x - x.mean()) / x.std()

# 左：y = r * x + sqrt(1 - r^2) * e で r = -0.4 としたデータ（負の相関）
y = -0.4 * x + np.sqrt(1 - 0.4**2) * e
y = (y - y.mean()) / y.std()

# 右：右上に外れ値を5個加える
x_out = np.array([3.0, 3.2, 3.4, 3.6, 3.8])
y_out = np.array([3.5, 3.1, 3.7, 3.3, 3.6])
x_all = np.concatenate([x, x_out])
y_all = np.concatenate([y, y_out])

r1 = np.corrcoef(x, y)[0, 1]
r2 = np.corrcoef(x_all, y_all)[0, 1]

plt.figure(figsize=(6, 3))

plt.subplot(1, 2, 1)
plt.scatter(x, y, color="#0050b3", s=10)
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.xlabel("x")
plt.ylabel("y")
plt.title(f"r = {r1:.2f}")

plt.subplot(1, 2, 2)
plt.scatter(x, y, color="#0050b3", s=10)
plt.scatter(x_out, y_out, color="#c81e1e", s=10)
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.xlabel("x")
plt.ylabel("y")
plt.title(f"r = {r2:.2f}")

plt.tight_layout()
plt.show()
```
