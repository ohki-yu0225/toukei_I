### 質的変数の集計

質的変数を要約するために，それぞれのカテゴリに属する観測値の個数（度数または頻度という）を集計する。

```python
freq = df["方角"].value_counts()
rel_freq = df["方角"].value_counts(normalize=True).round(3)
freq_table = pd.concat([freq, rel_freq], axis=1)
freq_table.columns = ["度数", "相対度数"]
freq_table
```

質的変数の集計結果は棒グラフで可視化できる。

```python
plt.bar(freq.index, freq, edgecolor="black")
plt.xlabel("方角")
plt.ylabel("度数")
plt.show()
```

```python
plt.bar(rel_freq.index, rel_freq, edgecolor="black")
plt.xlabel("方角")
plt.ylabel("相対度数")
plt.show()
```

### 度数分布表

量的変数の値をいくつかのグループに分けて，集計したものを度数分布という。

```python
bins = range(60000, 150001, 10000)
rent_freq, _ = np.histogram(df["家賃"], bins=bins)

labels = [f"{b}〜{b + 10000}" for b in bins[:-1]]

rent_table = pd.DataFrame({"度数": rent_freq, "相対度数": (rent_freq / rent_freq.sum()).round(3)}, index=labels)
rent_table
```

### ヒストグラム

度数分布をグラフ化する方法の1つにヒストグラムがある。ヒストグラムは横軸に変数の値を取り，それぞれの階級の区間上に面積と度数が比例するように長方形を描く。

```python
plt.hist(df["家賃"], bins=bins, edgecolor="black")
plt.xlabel("家賃（円）")
plt.ylabel("度数")
plt.show()
```

```python
plt.hist(df["家賃"], bins=bins, weights=np.ones(len(df)) / len(df), edgecolor="black")
plt.xlabel("家賃（円）")
plt.ylabel("相対度数")
plt.show()
```

### 分布の特徴の把握

ヒストグラムは量的変数の分布の特徴を把握することを目的としている。具体的には，分布の中心はどのあたりか，散らばりはどの程度の大きさか，全体として左右対称か，あるいはどちらかのすそが長い分布かなどの特徴を知ることができる。

```python
np.random.seed(0)
left = np.random.beta(5, 2, 1000) * 100
sym = np.random.beta(5, 5, 1000) * 100
right = np.random.beta(2, 5, 1000) * 100

fig, axes = plt.subplots(1, 3, figsize=(10, 3), sharex=True, sharey=True)
axes[0].hist(left, bins=10, edgecolor="black")
axes[0].set_title("左のすそが長い分布")
axes[1].hist(sym, bins=10, edgecolor="black")
axes[1].set_title("左右対称な分布")
axes[2].hist(right, bins=10, edgecolor="black")
axes[2].set_title("右のすそが長い分布")
plt.show()
```
